# Object Detection for Optic Disc and Macula

```python
from ultralytics import YOLO
import os
from PIL import Image, ImageDraw, ImageFont

# Load a pretrained YOLOv8n model
model = YOLO('best.pt')

# Define path to directory containing images and videos for inference
source = 'D:/gitvscode/odn/data/fundus/split4/images/val'

save_dir = 'D:/gitvscode/odn/data/fundus/split/detected_results'

# Make sure save directory exists
os.makedirs(save_dir, exist_ok=True)

# Perform prediction on each image in the source directory
for img_path in os.listdir(source):
    img_full_path = os.path.join(source, img_path)
    if os.path.isfile(img_full_path):
        # Perform inference
        results = model(img_full_path)

        # Retrieve the original image as a PIL image
        orig_img = results[0].orig_img
        img = Image.fromarray(orig_img)
        # Draw boxes on the image, if any
        if hasattr(results[0], 'boxes') and results[0].boxes.xyxy is not None:
            draw = ImageDraw.Draw(img)
            for box in results[0].boxes.xyxy:
                # Convert to a list of integers
                box = [int(coord) for coord in box.tolist()]
                # Outline boxes with a specified color and width
                draw.rectangle(box, outline='red', width=10)

                # Optionally, add labels
                cls = results[0].names[int(results[0].boxes.cls[0])]  # class label
                conf = results[0].boxes.conf[0]  # confidence
                label = f'{cls} {conf:.2f}'
                text_size = draw.textsize(label, font=ImageFont.truetype("arial.ttf", 30))
                text_position = (box[0], box[1] - text_size[1] - 10)
                # 绘制一个简单的文本背景
                draw.rectangle([text_position, (text_position[0] + text_size[0], text_position[1] + text_size[1])], fill='red')
                # 绘制文本
                draw.text(text_position, label, fill='white', font=ImageFont.truetype("arial.ttf", 30))

        # Save the image with drawn boxes
        save_path = os.path.join(save_dir, img_path)
        img.save(save_path)
        print(f"Saved predicted image to {save_path}")
```
